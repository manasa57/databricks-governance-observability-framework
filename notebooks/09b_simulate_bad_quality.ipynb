```python
from pyspark.sql import functions as F

customers = spark.table(
    "platform_monitoring.customers"
)

customers = customers.withColumn(

    "customer_name",

    F.when(
        F.rand() < .20,
        None
    )

    .otherwise(
        F.col("customer_name")
    )

)
```

```python
customers.write \
.mode("overwrite") \
.format("delta") \
.saveAsTable(
"platform_monitoring.customers"
)
```
