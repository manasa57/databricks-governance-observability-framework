```python
from pyspark.sql import functions as F

access_logs = spark.table(
    "platform_monitoring.access_logs"
)

access_logs = (

    access_logs

    .withColumn(

        "access_timestamp",

        F.when(
            F.col("table_name")=="audit_logs",
            F.current_timestamp() - F.expr("INTERVAL 120 DAYS")
        )

        .when(
            F.col("table_name")=="shipments",
            F.current_timestamp() - F.expr("INTERVAL 60 DAYS")
        )

        .otherwise(
            F.col("access_timestamp")
        )
    )
)
```

```python
access_logs.write \
.mode("overwrite") \
.format("delta") \
.saveAsTable(
"platform_monitoring.access_logs"
)
```
